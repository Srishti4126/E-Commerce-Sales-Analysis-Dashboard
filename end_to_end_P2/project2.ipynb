```python
import sqlite3


conn = sqlite3.connect('sales_database.db')

df.to_sql('orders', conn, if_exists='replace', index=False)

def sql(query):
    return pd.read_sql_query(query, conn)
```

```python
#Total number ofd orders?
sql("""
SELECT COUNT(*) AS total_orders
FROM orders;
""")
```

```python
#Total sales?
sql("""
SELECT SUM(net_amount) AS total_sales
FROM orders;
""")
```

```python
#Top 5 products?
sql("""
SELECT product, SUM(net_amount) AS sales
FROM orders
GROUP BY product
ORDER BY sales DESC
LIMIT 5;
""")
```

```python
#Top cities?
sql("""
SELECT city, SUM(net_amount) AS total_sales
FROM orders
GROUP BY city
ORDER BY total_sales DESC
LIMIT 5;
""")
```

```python
#Monthly Sales
sql("""
SELECT order_month, SUM(net_amount) AS total_sales
FROM orders
GROUP BY order_month;
""")
```

```python
#Highest Profit Products
sql("""
SELECT product, SUM(profit) AS total_profit
FROM orders
GROUP BY product
ORDER BY total_profit DESC
LIMIT 5;
""")
```

```python
#payment mode distribution
sql("""
SELECT payment_mode, COUNT(*) AS total_orders
FROM orders
GROUP BY payment_mode
ORDER BY total_orders DESC;
""")
```

```python
#cancelled orders
sql("""
SELECT COUNT(*) AS cancelled_orders
FROM orders
WHERE order_status = 'Cancelled';
""")
```
